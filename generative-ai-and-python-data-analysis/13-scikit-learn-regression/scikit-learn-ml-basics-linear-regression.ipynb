##### 프롬프트
> 2025년 5월 1일부터 20일 동안의 날짜를 기준으로, consumption, storage, oil_price, avg_temp, gas_price 데이터를 만들어서 pandas 데이터프레임으로 정리해줘. 리스트로 값도 같이 넣어줘.

> 2025년 5월 1일부터 시작해서 20일 동안의 가스 데이터를 만들고 싶어. 날짜(date), 소비량(consumption), 저장량(storage), 석유 가격(oil_price), 평균 기온(avg_temp), 가스 가격(gas_price) 컬럼을 가진 pandas 데이터프레임을 만들어줘.
- consumption은 리스트로 직접 주고,
- storage는 7000으로 고정된 값을 20개 만들고,
- oil_price는 20일간 약간씩 변동하는 숫자 리스트로 만들고,
- avg_temp도 10~14도 사이로 설정하고,
- gas_price는 예측 대상으로 적절한 숫자 리스트로 넣어줘.
마지막엔 데이터프레임을 출력하는 코드도 포함해줘.

```python
import pandas as pd

# 20일 동안의 날짜
dates = pd.date_range(start="2025-05-01", periods=20, freq='D')

# 가스 소비량
consumption = [15000, 15200, 14900, 14800, 15100, 15050, 15300, 14950, 15250, 15150,
               15000, 15300, 15200, 14900, 15050, 15100, 15200, 15300, 14950, 14850]

# 저장량 (7000 고정된 값)
storage = [7000] * 20

# 석유 가격 (1달러씩 변동)
oil_price = [78, 79, 77, 76, 80, 78, 79, 77, 76, 78,
             80, 81, 79, 78, 80, 82, 83, 81, 77, 76]

# 평균 기온
avg_temp = [10, 11, 12, 10, 11, 13, 14, 12, 13, 11,
            10, 12, 13, 14, 11, 10, 12, 13, 11, 10]

# 예측 대상 (가스 가격)
gas_price = [
    30.39, 30.33, 29.99, 29.44, 30.51,
    30.26, 30.58, 29.56, 30.37, 30.18,
    30.26, 31.03, 30.39, 29.68, 30.47,
    30.94, 31.11, 30.93, 29.86, 29.34
]

# 데이터프레임 생성
gas_data = pd.DataFrame({
    'date': dates,
    'consumption': consumption,
    'storage': storage,
    'oil_price': oil_price,
    'avg_temp': avg_temp,
    'gas_price': gas_price
})

# 출력
gas_data
```
