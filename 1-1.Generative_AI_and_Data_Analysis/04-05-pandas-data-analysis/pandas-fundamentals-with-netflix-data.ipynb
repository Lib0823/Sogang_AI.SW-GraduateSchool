* dataframe.agg({'열1': function1, '열2': function2, ...})
* 통계 함수들을 파이썬 딕셔너리 데이터 타입 형태로 전달할 수 있습니다.

```python
# 원하는 통계량 요약
print(netflix.agg({'release_year' : ['min', 'max', 'median', 'std'],
                   'duration' : ['min', 'max', 'mean', 'median']}))
```
