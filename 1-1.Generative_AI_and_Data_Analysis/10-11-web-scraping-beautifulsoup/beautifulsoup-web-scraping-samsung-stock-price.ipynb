#### **6. 데이터 확인을 위해 td 태그의 텍스트 값 출력**

슬라이드39.JPG

#### 프롬프트
> BeautifulSoup으로 첫 번째 <tr> 태그의 클래스 이름을 찾아서, 그 클래스가 붙은 <td> 태그들에서 텍스트 값을 순서대로 출력하는 코드를 작성해줘.

> BeautifulSoup 객체 soup에서 첫 번째 <tr> 태그를 찾아 클래스 이름을 추출한 뒤, 그 클래스 이름이 붙은 모든 <td> 태그 리스트에서 인덱스 0부터 7까지 텍스트 값을 각각 출력하는 파이썬 코드를 작성해줘. 출력하는 텍스트는 날짜, 시가, 고가, 저가, 종가, 조정 종가, 거래량 순서에 해당한다는 주석도 포함해줘.

```python
import requests
from bs4 import BeautifulSoup

stock_url = "https://finance.yahoo.com/quote/005930.KS/history/"
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/130.0.0.0 Safari/537.36',
    'accept': 'text/html,application/xhtml+xml,application/xml;q=0.9,image/avif,image/webp,image/apng,*/*;q=0.8,application/signed-exchange;v=b3;q=0.7'
    }
res = requests.get(stock_url, headers=headers)
html = res.text
soup = BeautifulSoup(html, "html.parser")
first_tr = soup.find('tr')
first_class = first_tr.get('class')[0]

# 데이터를 확인하기 위해 일부 td 태그의 텍스트 값을 출력
# 여기서는 각 데이터를 대표하는 첫 번째 행의 값을 출력
print(soup.find_all("td", class_=first_class)  ) # 날짜(Date)
print(soup.find_all("td", class_=first_class)  ) # 시가(Open)
print(soup.find_all("td", class_=first_class)  ) # 고가(High)
print(soup.find_all("td", class_=first_class)  ) # 저가(Low)
print(soup.find_all("td", class_=first_class)  ) # 종가(Close)
print(soup.find_all("td", class_=first_class)  ) # 조정 종가(Adj Close)
print(soup.find_all("td", class_=first_class)  ) # 거래량(Volume)
print(soup.find_all("td", class_=first_class)  )
# 다음 날짜(7개 단위로 다음 줄 데이터가 출력)
```
