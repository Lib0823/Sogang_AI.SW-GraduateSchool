#### 프롬프트
> seaborn과 matplotlib.pyplot을 사용해 ‘tips’ 데이터셋의 요일별 팁 금액을 성별로 구분하여 stripplot과 swarmplot 두 개를 각각 서브플롯에 그리는 코드를 작성해줘.

> 파이썬 seaborn과 matplotlib.pyplot 라이브러리를 불러와 ‘tips’ 데이터셋을 로드하고, figsize가 (15,5)인 figure에 1행 2열 서브플롯을 생성한 뒤, 첫 번째 서브플롯(ax1)에 ‘day’를 x축, ‘tip’을 y축으로, ‘sex’로 색 구분해 alpha 0.7인 stripplot을 그리고, 두 번째 서브플롯(ax2)에는 같은 축과 색 구분 기준으로 palette='Set2', alpha=0.7인 swarmplot을 그리며, 각각 제목을 ‘Strip Plot of Tip by Day and Gender’, ‘Swarm Plot of Tip by Day and Gender’로 설정하는 코드를 작성해줘.

```python
# 시본과 맷플롯립 라이브러리 불러오기
import seaborn as sns
import matplotlib.pyplot as plt

# 팁 데이터셋 불러오기
tips = sns.load_dataset('tips')

# figure에 2개의 서브 플롯을 생성
fig = plt.figure(figsize=(15, 5))
ax1 = fig.add_subplot(1, 2, 1)
ax2 = fig.add_subplot(1, 2, 2)

# stripplot( ) 그리기
sns.stripplot(x='day', y='tip', hue='sex', data=tips, alpha=0.7, ax=ax1)

# swarmplot( ) 그리기


# 서브 플롯의 제목 설정
ax1.set_title('Strip Plot of Tip by Day and Gender')
ax2.set_title('Swarm Plot of Tip by Day and Gender')
plt.show()
```

#### 프롬프트
> seaborn과 matplotlib.pyplot을 사용해 ‘tips’ 데이터셋에서 ‘total_bill’과 ‘tip’의 산점도를 그리고, 한 그래프에는 선형 회귀선을 표시하고 다른 그래프에는 표시하지 않는 두 개의 서브플롯을 만드는 코드를 작성해줘.

> 파이썬 seaborn과 matplotlib.pyplot 라이브러리를 불러와 ‘tips’ 데이터셋을 로드하고, figsize (15, 5)의 figure에 1행 2열 서브플롯을 만든 뒤, 첫 번째 서브플롯(ax1)에 ‘total_bill’과 ‘tip’의 산점도를 점 크기 50, 투명도 0.5로 그리고 파란색 점선 형태의 선형 회귀선(fit_reg=True)을 추가하며, 두 번째 서브플롯(ax2)에는 동일한 산점도를 그리고 선형 회선은 표시하지 않는(fit_reg=False) 코드를 작성해줘. 전체 제목은 ‘Scatter Plots with Regression Lines’로 설정하고 각 서브플롯에 ‘fit_reg = True’, ‘fit_reg = False’ 제목을 붙여줘.



```python
# 시본과 맷플롯립 라이브러리 불러오기
import seaborn as sns
import matplotlib.pyplot as plt

# 팁 데이터셋 불러오기
tips = sns.load_dataset('tips')

# figure에 2개의 서브 플롯을 생성
fig = plt.figure(figsize=(15, 5))
ax1 = fig.add_subplot(1, 2, 1)
ax2 = fig.add_subplot(1, 2, 2)

# 산점도에 선형 회귀선 표시(fit_reg=True)
sns.regplot(x='total_bill', y='tip', data=tips, color='blue',
            scatter_kws={'s': 50, 'alpha': 0.5}, line_kws={'linestyle': '--'}, ax=ax1)

# 산점도에 선형 회귀선 미표시(fit_reg=False)
sns.regplot(x='total_bill', y='tip', data=tips, color='blue',
            scatter_kws={'s': 50, 'alpha': 0.5}, line_kws={'linestyle': '--'}
            ,              )

# 제목 설정
fig.suptitle('Scatter Plots with Regression Lines', fontsize=16)
ax1.set_title('fit_reg = True')
ax2.set_title('fit_reg = False')
plt.show()
```
